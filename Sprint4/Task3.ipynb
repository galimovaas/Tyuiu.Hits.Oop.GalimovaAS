Описание задачи:
Создать базовый класс Product в C#, который будет представлять информацию о
продуктах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Product:
• Атрибуты: Название (Name), Цена (Price), Производитель (Manufacturer).
• Методы:
o GetInfo(): метод для получения информации о продукте в виде строки.
o Discount(): метод для применения скидки к цене продукта.
o Display(): метод для отображения информации о продукте на экране.
Требования к производным классам:
1. Электроника (Electronics): Должен содержать дополнительные атрибуты,
такие как Гарантийный срок (WarrantyPeriod). Метод Discount() должен быть
переопределен для добавления логики учета гарантийного срока при
применении скидки.
2. Одежда (Clothing): Должен содержать дополнительные атрибуты, такие как
Размер (Size). Метод Display() должен быть переопределен для добавления
информации о размере при отображении информации о продукте.
3. Книги (Books) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Автор (Author). Метод GetInfo() должен
быть переопределен для включения информации об авторе в описании
продукта.

```

public interface IProduct
{
    string GetInfo();
    void ApplyDiscount(double percentage);
    void Display();
}

public interface IInventory
{
    void UpdateStock(int quantity);
    int GetStockQuantity();
}

public abstract class Product : IProduct, IInventory
{
    public string Name { get; set; }
    public decimal Price { get; set; }
    public string Manufacturer { get; set; }
    public string ProductId { get; set; }
    public int StockQuantity { get; protected set; }
    public string Category { get; set; }
    public double Weight { get; set; }

    protected Product(string name, decimal price, string manufacturer)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
        ProductId = $"PRD_{DateTime.Now.Ticks}";
        StockQuantity = 0;
    }

    public virtual string GetInfo()
    {
        return $"Продукт: {Name}\nЦена: {Price:C}\nПроизводитель: {Manufacturer}\nID: {ProductId}";
    }

    public virtual void ApplyDiscount(double percentage)
    {
        if (percentage < 0 || percentage > 100)
            throw new ArgumentException("Скидка должна быть между 0 и 100%");
        
        Price -= Price * (decimal)(percentage / 100);
    }

    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }

    public virtual double CalculateShippingCost()
    {
        return Weight * 50;
    }

    public virtual void UpdateStock(int quantity)
    {
        StockQuantity += quantity;
    }

    public virtual int GetStockQuantity()
    {
        return StockQuantity;
    }

    public void Restock(int quantity)
    {
        UpdateStock(quantity);
        Console.WriteLine($"Товар {Name} пополнен. Текущий остаток: {StockQuantity}");
    }
}

public class Electronics : Product
{
    public int WarrantyPeriod { get; set; }
    public string Model { get; set; }
    public bool HasBattery { get; set; }

    public Electronics(string name, decimal price, string manufacturer, int warrantyPeriod) 
        : base(name, price, manufacturer)
    {
        WarrantyPeriod = warrantyPeriod;
        Category = "Электроника";
    }

    public override void ApplyDiscount(double percentage)
    {
        double adjustedPercentage = WarrantyPeriod > 24 ? percentage * 0.7 : percentage;
        base.ApplyDiscount(adjustedPercentage);
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $"\nГарантия: {WarrantyPeriod} месяцев\nМодель: {Model}";
    }

    public override double CalculateShippingCost()
    {
        return base.CalculateShippingCost() * (HasBattery ? 1.3 : 1);
    }

    public void InstallSoftware(string softwareName)
    {
        Console.WriteLine($"Установка {softwareName} на {Name}");
    }
}

public class Clothing : Product
{
    public string Size { get; set; }
    public string Color { get; set; }
    public string Material { get; set; }

    public Clothing(string name, decimal price, string manufacturer, string size) 
        : base(name, price, manufacturer)
    {
        Size = size;
        Category = "Одежда";
        Weight = 0.5;
    }

    public override void Display()
    {
        base.Display();
        Console.WriteLine($"Размер: {Size}\nЦвет: {Color}\nМатериал: {Material}");
    }

    public string GetCareInstructions()
    {
        return $"Инструкция по уходу: стирка при 30°C";
    }
}

public class Books : Product
{
    public string Author { get; set; }
    public int PageCount { get; set; }
    public string Genre { get; set; }

    public Books(string name, decimal price, string manufacturer, string author) 
        : base(name, price, manufacturer)
    {
        Author = author;
        Category = "Книги";
        Weight = 0.3;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $"\nАвтор: {Author}\nЖанр: {Genre}\nСтраниц: {PageCount}";
    }

    public override void ApplyDiscount(double percentage)
    {
        base.ApplyDiscount(Math.Min(percentage, 30));
    }

    public string GetReadingTime()
    {
        return PageCount <= 300 ? "Быстрое чтение" : "Длительное чтение";
    }
}

public class ProductService
{
    private readonly List<Product> _products = new List<Product>();
    
    public void AddProduct(Product product)
    {
        _products.Add(product);
    }
    
    public IEnumerable<Product> GetAllProducts()
    {
        return _products;
    }
    
    public decimal CalculateTotalValue()
    {
        return _products.Sum(p => p.Price * p.GetStockQuantity());
    }
}


        var productService = new ProductService();
        
        var laptop = new Electronics("Ноутбук Gaming", 150000, "ASUS", 24)
        {
            Model = "ROG Strix G15",
            HasBattery = true,
            Weight = 2.5
        };
        laptop.UpdateStock(10);

        var tshirt = new Clothing("Футболка", 2500, "Nike", "L")
        {
            Color = "Черный",
            Material = "Хлопок"
        };
        tshirt.UpdateStock(50);

        var book = new Books("Война и мир", 1500, "АСТ", "Лев Толстой")
        {
            PageCount = 1225,
            Genre = "Роман"
        };
        book.UpdateStock(25);

        productService.AddProduct(laptop);
        productService.AddProduct(tshirt);
        productService.AddProduct(book);

        
        foreach (var product in productService.GetAllProducts())
        {
            product.Display();
            Console.WriteLine($"Доставка: {product.CalculateShippingCost():C}");
            Console.WriteLine($"На складе: {product.GetStockQuantity()}");
            Console.WriteLine("---");
        }

        Console.WriteLine("\n=== ПРИМЕНЕНИЕ СКИДОК ===\n");
        
        laptop.ApplyDiscount(15);
        tshirt.ApplyDiscount(20);
        book.ApplyDiscount(40);
        
        Console.WriteLine($"Ноутбук: {laptop.Price:C}");
        Console.WriteLine($"Футболка: {tshirt.Price:C}");
        Console.WriteLine($"Книга: {book.Price:C}");

        
        laptop.InstallSoftware("Office");
        Console.WriteLine(tshirt.GetCareInstructions());
        Console.WriteLine(book.GetReadingTime());

        Console.WriteLine($"\nОбщая стоимость: {productService.CalculateTotalValue():C}");

        Console.WriteLine("\n=== ПОПОЛНЕНИЕ ЗАПАСОВ ===\n");
        laptop.Restock(5);
```
